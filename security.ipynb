```python
import osmnx as ox
import pandas as pd
import warnings

# Tell OSMnx not to create a local cache folder
ox.settings.use_cache = False 
warnings.filterwarnings("ignore") # Hides standard coordinate math warnings

# 1. Define the exact geographic boundary you want to scrape
place_name = "Delhi, India"

# 2. Tell OSMnx exactly what infrastructure to look for
# OpenStreetMap tags police stations as amenities
tags = {'amenity': 'police'}

# 3. Download the data (this usually takes 15 to 45 seconds)
print(f"Reaching out to OpenStreetMap servers for {place_name}...")
police_stations = ox.features_from_place(place_name, tags)

# 4. Clean the data to extract the exact GPS coordinates
# OSMnx returns building footprints (polygons). .centroid finds the exact center point.
police_stations['Latitude'] = police_stations.geometry.centroid.y
police_stations['Longitude'] = police_stations.geometry.centroid.x

# Keep only the name of the station and its coordinates
clean_data = police_stations[['name', 'Latitude', 'Longitude']].dropna()

# 5. Save it immediately as a real, usable CSV file in your VS Code folder
clean_data.to_csv("delhi_real_police_stations.csv", index=False)

print(f"Success! Downloaded coordinates for {len(clean_data)} real police stations.")
print(clean_data.head())
```

```python
import osmnx as ox
import pandas as pd
import warnings

ox.settings.use_cache = False 
warnings.filterwarnings("ignore")

place_name = "Delhi, India"
tags_hospital = {'amenity': 'hospital'}

print("Reaching out to OpenStreetMap servers for Delhi Hospitals...")
hospitals = ox.features_from_place(place_name, tags_hospital)

hospitals['Latitude'] = hospitals.geometry.centroid.y
hospitals['Longitude'] = hospitals.geometry.centroid.x

clean_hospitals = hospitals[['name', 'Latitude', 'Longitude']].dropna()
clean_hospitals.to_csv("delhi_real_hospitals.csv", index=False)

print(f"Success! Downloaded coordinates for {len(clean_hospitals)} real hospitals.")
print(clean_hospitals.head())
```
